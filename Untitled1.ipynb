```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 生成一些假数据：y = 2x + 1 + 噪声
np.random.seed(42)
X = np.random.rand(100, 1) * 10
y = 2 * X + 1 + np.random.randn(100, 1) * 2

# 训练线性回归模型
model = LinearRegression()
model.fit(X, y)

# 打印学到的参数
print(f"斜率（权重）：{model.coef_[0][0]:.2f}")
print(f"截距（偏置）：{model.intercept_[0]:.2f}")

# 画图
plt.scatter(X, y, alpha=0.6, label='数据')
plt.plot(X, model.predict(X), color='red', label='拟合直线')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.show()
```
